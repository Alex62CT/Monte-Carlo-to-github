```python
%%bash
set -e # Exit immediately if a command exits with a non-zero status.

ROOT_SOURCE_DIR="root" # This is where the cloned source code will reside
ROOT_BUILD_DIR="$ROOT_SOURCE_DIR/build"
ROOT_INSTALL_PREFIX="/content/root-install"

# Check if ROOT is already installed
if [ -d "$ROOT_INSTALL_PREFIX/bin" ] && [ -f "$ROOT_INSTALL_PREFIX/bin/root" ]; then
    echo "ROOT is already installed at $ROOT_INSTALL_PREFIX. Skipping build."
    exit 0 # Exit successfully
fi

echo "ROOT not found, proceeding with fresh build."

# Ensure a clean slate for the ROOT source directory if it exists, otherwise clone
if [ -d "$ROOT_SOURCE_DIR" ]; then
    echo "Removing existing ROOT source directory: $ROOT_SOURCE_DIR"
    rm -rf "$ROOT_SOURCE_DIR"
fi

echo "Cloning ROOT (shallow)..."
time git clone --depth 1 https://github.com/root-project/root.git "$ROOT_SOURCE_DIR"
echo "ROOT repository cloned."

# --- Debugging additions ---
echo "Contents of /$ROOT_SOURCE_DIR after cloning (should contain CMakeLists.txt and other source files):"
ls -F "/$ROOT_SOURCE_DIR"
# --- End Debugging additions ---

# Dependencies are assumed to be handled by prior cells (e.g., G83zgFLBgaoj).
# Removed redundant apt-get install from here.

# Build directory
mkdir -p "$ROOT_BUILD_DIR"
cd "$ROOT_BUILD_DIR"
echo "Changed to $ROOT_BUILD_DIR directory."

# No need to rm -rf * here if source dir is cleaned at the beginning.
# If source changes, we re-clone. If build directory gets stale, CMake will usually handle it.

echo "Configuring ROOT (minimal build)..."
time cmake .. \
  -DCMAKE_INSTALL_PREFIX="$ROOT_INSTALL_PREFIX" \
  -DCMAKE_CXX_STANDARD=17 \
  -Dgminimal=ON \
  -Droofit=OFF \
  -Dtmva=OFF \
  -Dmathmore=OFF \
  -Dx11=OFF \
  -Dopengl=OFF \
  -Dpyroot=OFF \
  -Dpython=OFF \
  -Dtpython=OFF \
  -Dtests=OFF \
  -Dexamples=OFF
echo "ROOT configured."

echo "Building ROOT..."
time make -j$(nproc)
echo "ROOT built."

echo "Installing ROOT..."
time make install
echo "ROOT installed."

echo "Checking installation in $ROOT_INSTALL_PREFIX..."
if [ -d "$ROOT_INSTALL_PREFIX/bin" ]; then
    ls -l "$ROOT_INSTALL_PREFIX/bin"
    echo "ROOT installation verified successfully."
else
    echo "Error: $ROOT_INSTALL_PREFIX/bin not found after installation."
    exit 1
fi
```

### Build ROOT from Source

This cell will clone the ROOT project, install necessary dependencies, configure the build, compile ROOT, and install it to `/content/root-install`. This ensures a robust installation compatible with the environment.

```python
%%bash
set -e # Exit immediately if a command exits with a non-zero status.

ROOT_SOURCE_DIR="root" # This is where the cloned source code will reside
ROOT_BUILD_DIR="$ROOT_SOURCE_DIR/build"
ROOT_INSTALL_PREFIX="/content/root-install"

# Always ensure a clean slate for the ROOT source directory
if [ -d "$ROOT_SOURCE_DIR" ]; then
    echo "Removing existing ROOT source directory: $ROOT_SOURCE_DIR"
    rm -rf "$ROOT_SOURCE_DIR"
fi

echo "Cloning ROOT (shallow)..."
git clone --depth 1 https://github.com/root-project/root.git "$ROOT_SOURCE_DIR"
echo "ROOT repository cloned."

# --- Debugging additions ---
echo "Contents of /$ROOT_SOURCE_DIR after cloning (should contain CMakeLists.txt and other source files):"
ls -F "/$ROOT_SOURCE_DIR"
# --- End Debugging additions ---

# Dipendenze minime
echo "Installing dependencies..."
apt-get update -qq # Use -qq for quiet update
apt-get install -y -qq \
  build-essential cmake \
  libx11-dev libxpm-dev libxft-dev \
  libxext-dev
echo "Dependencies installed."

# Build directory
mkdir -p "$ROOT_BUILD_DIR"
cd "$ROOT_BUILD_DIR"
echo "Changed to $ROOT_BUILD_DIR directory."

# Clean build directory (just in case)
echo "Cleaning $ROOT_BUILD_DIR directory..."
rm -rf *
echo "Build directory cleaned."

# --- Debugging additions ---
echo "Current directory before CMake:"
pwd
# --- End Debugging additions ---

echo "Configuring ROOT (minimal build)..."
# CMake source directory is one level up from build directory
cmake .. \
  -DCMAKE_INSTALL_PREFIX="$ROOT_INSTALL_PREFIX" \
  -DCMAKE_CXX_STANDARD=17 \
  -Dgminimal=ON \
  -Droofit=OFF \
  -Dtmva=OFF \
  -Dmathmore=OFF \
  -Dx11=OFF \
  -Dopengl=OFF \
  -Dpyroot=OFF \
  -Dpython=OFF \
  -Dtpython=OFF \
  -Dtests=OFF \
  -Dexamples=OFF
echo "ROOT configured."

echo "Building ROOT..."
make -j$(nproc)
echo "ROOT built."

echo "Installing ROOT..."
make install
echo "ROOT installed."

echo "Checking installation in $ROOT_INSTALL_PREFIX..."
if [ -d "$ROOT_INSTALL_PREFIX/bin" ]; then
    ls -l "$ROOT_INSTALL_PREFIX/bin"
    echo "ROOT installation verified successfully."
else
    echo "Error: $ROOT_INSTALL_PREFIX/bin not found after installation."
    exit 1
fi
```
